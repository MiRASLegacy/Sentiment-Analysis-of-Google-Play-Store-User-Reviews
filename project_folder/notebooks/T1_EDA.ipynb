# Task 1: Exploratory Data Analysis
**Objective:** Clean the Google Play Store User Reviews dataset and perform initial visual analysis.
**Inputs:** `../data/raw/googleplaystore_user_reviews.csv`
**Outputs:** `../data/cleaned.csv`

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/googleplaystore_user_reviews.csv')

df.dropna(inplace=True)
df.drop_duplicates(inplace=True)

display(df.head())
display(df.info())
```

### 1. Sentiment Distribution
**Question:** What is the ratio of positive, negative, and neutral reviews in the dataset?

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Sentiment', order=df['Sentiment'].value_counts().index)
plt.title('Distribution of Sentiments')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.savefig('../reports/sentiment_distribution.png', bbox_inches='tight')
plt.show()
```

### 2. Sentiment Polarity Distribution
**Question:** How is the sentiment polarity distributed?

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Sentiment_Polarity'], bins=30, kde=True)
plt.title('Distribution of Sentiment Polarity')
plt.xlabel('Polarity')
plt.ylabel('Frequency')
plt.show()
```

### 3. Sentiment Subjectivity Distribution
**Question:** How subjective are the user reviews on average?

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Sentiment_Subjectivity'], bins=30, kde=True)
plt.title('Distribution of Sentiment Subjectivity')
plt.xlabel('Subjectivity')
plt.ylabel('Frequency')
plt.show()
```

### 4. Polarity vs Subjectivity
**Question:** Is there a correlation between the subjectivity of a review and its polarity?

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Sentiment_Subjectivity', y='Sentiment_Polarity', alpha=0.5)
plt.title('Sentiment Polarity vs Subjectivity')
plt.xlabel('Subjectivity')
plt.ylabel('Polarity')
plt.show()
```

### 5. Polarity Range by Sentiment Class
**Question:** What is the range and presence of outliers in polarity values for each sentiment class?

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Sentiment', y='Sentiment_Polarity')
plt.title('Sentiment Polarity by Sentiment Class')
plt.xlabel('Sentiment')
plt.ylabel('Polarity')
plt.show()
```
